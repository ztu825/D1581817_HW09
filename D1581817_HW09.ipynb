## 題目說明
本作業利用 Python 整理「桌遊體驗」與「電影欣賞」兩場活動的報名資料。主要流程包括：
1. 移除多餘空白、統一轉小寫，並以逗號拆分名單（保留原始重複順序）。
2. 使用集合（Set）進行去重與集合運算（交集、聯集、差集）。
3. 利用排序與切片取出預覽通知名單。
4. 使用字典（Dict）保存人數與費用摘要，並輸出完整報表。

## 測試一：原始資料

```python
# =============================================================================
# 測試一：原始資料輸入
# =============================================================================
# 固定的兩個活動名稱，以 tuple 保存。
activity_names = ("桌遊體驗", "電影欣賞")

# 收到的原始文字名單，請保留原值。
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 用活動名稱查詢每人費用。
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# =============================================================================
# 1. 讀取活動名稱，建立兩份整理後的串列 (保留重複、轉小寫、去空白)
# =============================================================================
board_activity = activity_names[0]
movie_activity = activity_names[1]

# 清理字串：先移除所有半形空白，再轉換成小寫，最後依逗號分割
board_clean_str = board_raw.replace(" ", "").lower()
board_list = board_clean_str.split(",")

movie_clean_str = movie_raw.replace(" ", "").lower()
movie_list = movie_clean_str.split(",")

# =============================================================================
# 2. 使用集合去重並比較名單
# =============================================================================
board_set = set(board_list)
movie_set = set(movie_list)

# 共同參加者 (交集)
common_set = board_set & movie_set
common_list = sorted(list(common_set))

# 全部參加者 (聯集)
all_set = board_set | movie_set
all_list = sorted(list(all_set))

# 只參加桌遊者 (差集)
only_board_set = board_set - movie_set
only_board_list = sorted(list(only_board_set))

# 只參加電影者 (差集)
only_movie_set = movie_set - board_set
only_movie_list = sorted(list(only_movie_set))

# 各自去重排序名單
board_unique_sorted = sorted(list(board_set))
movie_unique_sorted = sorted(list(movie_set))

# =============================================================================
# 3. 建立排序串列，使用切片取得前兩位
# =============================================================================
notification_preview = all_list[:2]

# =============================================================================
# 4. 使用字典保存摘要，計算人數與應收費用
# =============================================================================
summary = {
    "board_count": len(board_set),
    "movie_count": len(movie_set),
    "common_count": len(common_set),
    "total_unique_count": len(all_set),
    "total_registrations": len(board_set) + len(movie_set),
    "board_fee": len(board_set) * unit_fees[board_activity],
    "movie_fee": len(movie_set) * unit_fees[movie_activity],
}
summary["total_fee"] = summary["board_fee"] + summary["movie_fee"]

# =============================================================================
# 5. 輸出完整摘要
# =============================================================================
print("HW09 活動報名摘要 (測試一)")
print(f"活動：{board_activity}，每人 {unit_fees[board_activity]:.2f} 元")
print(f"活動：{movie_activity}，每人 {unit_fees[movie_activity]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_unique_sorted}")
print(f"電影去重名單：{movie_unique_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{common_list}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{only_board_list}")
print(f"只參加電影者：{only_movie_list}")
print(f"全部參加者：{all_list}")
print(f"總不重複人數：{summary['total_unique_count']} 人")
print(f"報名人次：{summary['total_registrations']} 人次")
print(f"通知名單前兩位：{notification_preview}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```

### 問題一： 測試一中，整理後的桌遊串列有 4 筆資料，為什麼桌遊費用應以 3 人計算？直接把字串拆成 list，和再轉成 set，各自保留了什麼、失去了什麼？

- **回答提示**：
  1. **計費人數**：桌遊整理後的串列為 `['amy', 'ben', 'amy', 'cora']`。因為 `amy` 重複報名，主辦規則規定「同一個人在同一場活動中重複出現，只算一位參加者，也只收一次該場費用」，因此去重後的名單為 `['amy', 'ben', 'cora']` 共 3 人，應收 3 人的費用（120 * 3 = 360 元）。
  2. **List 的特性**：
     - **保留了**：元素的「原始順序」以及「重複的元素」。
     - **失去了**：快速判斷不重複成員的能力（在進行名單比對時效率較低）。
  3. **Set 的特性**：
     - **保留了**：唯一的、不重複的成員元素。
     - **失去了**：元素的「順序性」（集合是無序的，且會遺失原來的重複次數資訊）。

### 問題二： 測試三中，兩場活動只有 3 位不同的人，為什麼是 6 人次、總應收仍是 600 元？請用集合關係與算式說明。

- **回答提示**：
  1. **集合關係**：
     - 在測試三中，桌遊集合 $A = \{\text{amy}, \text{ben}, \text{cora}\}$，電影集合 $B = \{\text{amy}, \text{ben}, \text{cora}\}$。
     - 聯集（總不重複人數）為 $A \cup B = \{\text{amy}, \text{ben}, \text{cora}\}$，共 3 人。
     - 交集（兩場皆參加者）為 $A \cap B = \{\text{amy}, \text{ben}, \text{cora}\}$，共 3 人。
  2. **人次與費用算式**：
     - **報名人次**：指兩場活動各自去重後人數的相加。公式為 $|A| + |B| = 3 + 3 = 6$ 人次。這代表這 3 個人兩場活動都報名了，共產生 6 個活動名額。
     - **費用計算**：由於每位參加者參加不同活動是分別計費的。桌遊費用為 $3 \times 120 = 360$ 元，電影費用為 $3 \times 80 = 240$ 元。總應收為 $360 + 240 = 600$ 元。
